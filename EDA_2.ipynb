```python
def conv(values):
  if values==1:
    return "yes"
  else:
    return "no"
```

```python
df["SeniorCitizen"]=df["SeniorCitizen"].apply(conv)
df.head()
```

```python
df.info()
```

```python
ax=sns.countplot(x="Churn",data=df)
ax.bar_label(ax.containers[0])
plt.title("Count of customer by Churn")
plt.show()
```

```python
plt.figure(figsize=(3,4))
gb=df.groupby("Churn").agg({'Churn':"count"})
plt.pie(gb['Churn'],labels=gb.index,autopct="%1.2f%%")
plt.title("Percent of customer by Churn",fontsize=10)
plt.show()
```

```python
plt.figure(figsize=(4,5))
ax1=sns.countplot(x="gender",data=df)
ax1.bar_label(ax1.containers[0])
plt.title("count of customer by gender")
plt.show()
```

```python
plt.figure(figsize=(4,4))
ax2=sns.countplot(x="SeniorCitizen",data=df,hue="Churn")
ax2.bar_label(ax2.containers[0])
plt.show()
```

```python
#List of categorical columns
cols = ['PhoneService', 'MultipleLines', 'InternetService',
        'OnlineSecurity', 'OnlineBackup', 'DeviceProtection',
        'TechSupport', 'StreamingTV', 'StreamingMovies']

# Set up subplot grid (3 rows × 3 columns for 9 plots)
fig, axes = plt.subplots(nrows=3, ncols=3, figsize=(15, 12))
axes = axes.flatten()  # Flatten to easily iterate

# Loop through columns and axes
for i, col in enumerate(cols):
    ax = axes[i]
    sns.countplot(data=df, x=col, ax=ax)
    ax.set_title(col)
    ax.tick_params(axis='x', rotation=45)

    # Add labels on top of bars
    for container in ax.containers:
        ax.bar_label(container, label_type='edge', padding=3)

# Adjust layout
plt.tight_layout()
plt.show()
```
